# 02 — Exploratory Data Analysis

Synthetic healthcare portfolio dataset only.

```python
import pandas as pd, matplotlib.pyplot as plt, seaborn as sns
from pathlib import Path
df=pd.read_csv(Path('../data/healthcare_python_cleaned.csv'),parse_dates=['VisitDate'])
```

```python
display(df.describe(include='all').T); display(df['Diagnosis'].value_counts())
```

```python
sns.histplot(data=df,x='Age',kde=True); plt.title('Age distribution'); plt.show()
```

```python
sns.boxplot(data=df,x='Diagnosis',y='BMI'); plt.xticks(rotation=30); plt.show()
```

```python
monthly=df.set_index('VisitDate').resample('MS')['PatientID'].count()
monthly.plot(marker='o',figsize=(10,4)); plt.title('Monthly encounter records'); plt.show()
```

```python
sns.heatmap(df[['Age','BMI','SystolicBP','DiastolicBP','FastingGlucose','LengthOfStayDays']].corr(),annot=True); plt.show()
```
